Calculate monthly revenue. Given a list of daily sales, calculate total revenue for the month using list comprehension. Input: [1200, 1500, 980, 2100, 1300, 1800, 900]

```python
daily_sales = [1200, 1500, 980, 2100, 1300, 1800, 900]
total_revenue = 0
for sale in daily_sales:
    total_revenue += sale
print("Total monthly revenue:", total_revenue)
```

Website analytics finds peak traffic hour. Given hourly traffic numbers, find the hour with max traffic using a loop (no max() function). Input: [150, 320, 280, 450, 210, 380, 290, 500, 180]

```python
hourly_traffic = [150, 320, 280, 450, 210, 380, 290, 500, 180]
max_traffic = hourly_traffic[0]
peak_hour = 0

for hour, traffic in enumerate(hourly_traffic):
    if traffic > max_traffic:
        max_traffic = traffic
        peak_hour = hour

print(f"Peak traffic hour: {peak_hour} (Hour {peak_hour} counted from 0)")
print(f"Peak traffic: {max_traffic}")
```

Construction company calculates land plots. Create a Rectangle class with length and width. Add methods for area() and perimeter(). Test with 5x10 plot.

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
    
    def area(self):
        return self.length * self.width
    
    def perimeter(self):
        return 2 * (self.length + self.width)

# Test with 5x10 plot
plot = Rectangle(5, 10)
print(f"Area: {plot.area()}")
print(f"Perimeter: {plot.perimeter()}")
```

Bank creates account management system. Build BankAccount class with holder name and balance. Implement deposit() and withdraw(). Perform: deposit 1000, withdraw 200, deposit 500.

```python
class BankAccount:
    def __init__(self, holder_name, balance=0):
        self.holder_name = holder_name
        self.balance = balance
    
    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. New balance: {self.balance}")
    
    def withdraw(self, amount):
        if amount > self.balance:
            print(f"Insufficient funds. Current balance: {self.balance}")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")

# Test the account
account = BankAccount("John Doe")
account.deposit(1000)
account.withdraw(200)
account.deposit(500)
```
