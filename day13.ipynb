3. Create a 4×14 matrix (4 batsmen × 14 matches) with all runs

TASK 2.2: Performance Trends
1. Calculate 3-match moving average for each player's runs
2. Identify matches where player scored > mean + 1 std (exceptional performances)
3. Find correlation between balls_faced and runs_scored for all batsmen

```python
import numpy as np

def analyze_player(player):
    runs = np.array(player["runs_scored"])

    mean_runs = np.mean(runs)
    std_runs = np.std(runs)
    median_runs = np.median(runs)
    max_runs = np.max(runs)
    min_runs = np.min(runs)

    # consistency score
    if std_runs == 0:
        consistency = 0
    else:
        consistency = mean_runs / std_runs

    #return mean_runs, std_runs, median_runs, max_runs, min_runs, consistency

    return {
        "mean": round(mean_runs, 2),
        "std": round(std_runs, 2),
        "median": median_runs,
        "highest": max_runs,
        "lowest": min_runs,
        "consistency": round(consistency, 2)
    }


# 🔥 RUN FOR BOTH
for player_id, player in players.items():

    role = player["role"].lower().replace(" ", "-")

    if role in ["batsman", "all-rounder"]:

        stats = analyze_player(player)

        print(f"\n{player['name']} ({player['role']})")
        print(f"Mean Runs: {stats['mean']}")
        print(f"Std Dev: {stats['std']}")
        print(f"Median: {stats['median']}")
        print(f"Highest: {stats['highest']}")
        print(f"Lowest: {stats['lowest']}")
        print(f"Consistency Score: {stats['consistency']}")
```

```python
# Collect runs of valid players
runs_matrix_list = []
player_names = []

for player_id, player in players.items():
    role = player["role"].lower().replace(" ", "-")

    if role in ["batsman", "all-rounder"]:
        runs_matrix_list.append(player["runs_scored"])
        player_names.append(player["name"])

# Convert to NumPy 2D array (4 × 14)
runs_matrix = np.array(runs_matrix_list)

print("\nRuns Matrix (7 players × 14 matches):")
print(len(runs_matrix_list)) 
print(runs_matrix)
```

TASK 2.2: Performance Trends
1. Calculate 3-match moving average for each player's runs
2. Identify matches where player scored > mean + 1 std (exceptional performances)
3. Find correlation between balls_faced and runs_scored for all batsmen

```python
def moving_average(runs, window=3):
    runs_array = np.array(runs)
    return np.convolve(runs_array, np.ones(window)/window, mode='valid')

print("\n--- 3 Match Moving Average ---")

for player_id, player in players.items():
    role = player["role"].lower().replace(" ", "-")

    if role in ["batsman", "all-rounder"]:
        ma = moving_average(player["runs_scored"])

        print(f"\n{player['name']}:")
        print(np.round(ma, 2))

print("\n--- Exceptional Performances ---")

for player_id, player in players.items():
    role = player["role"].lower().replace(" ", "-")

    if role in ["batsman", "all-rounder"]:
        runs = np.array(player["runs_scored"])

        mean = np.mean(runs)
        std = np.std(runs)

        threshold = mean + std

        exceptional_matches = np.where(runs > threshold)[0]

        print(f"\n{player['name']}:")
        print(f"Threshold: {round(threshold, 2)}")

        if len(exceptional_matches) == 0:
            print("No exceptional performances")
        else:
            for idx in exceptional_matches:
                print(f"Match {idx+1}: {runs[idx]} runs")


print("\n--- Correlation (Balls Faced vs Runs) ---")

for player_id, player in players.items():
    role = player["role"].lower().replace(" ", "-")

    if role in ["batsman", "all-rounder"]:
        runs = np.array(player["runs_scored"])
        balls = np.array(player["balls_faced"])

        correlation = np.corrcoef(runs, balls)[0, 1]

        print(f"{player['name']}: {round(correlation, 2)}")

```
