EXERCISE 1: STUDENT GRADE MANAGEMENT SYSTEM

PROBLEM:
Create a student grade tracking system that stores data and generates reports.

REQUIREMENTS:

1. Data Structure:
   Create a dictionary to store students with this structure:
   {
       "S001": {
           "name": "Alice",
           "grades": [85, 90, 78, 92],  # list of test scores
           "attendance": 0.90,  # percentage as decimal (0-1)
           "projects": {"proj1": 95, "proj2": 88}
       }
   }

2. Functions to Write:
   - add_student(students, student_id, name)
   - add_grade(students, student_id, grade)
   - calculate_test_average(students, student_id)
   - calculate_final_score(students, student_id)
     Formula: Tests 60% + Projects 25% + Attendance 15%
     (Attendance score = attendance * 100)
   - get_letter_grade(score)
     A: 90+, B: 80-89, C: 70-79, D: 60-69, F: <60
   - save_to_file(students, filename)
   - load_from_file(filename)

3. Create Data:
   - Add at least 5 students with complete data
   - Each student should have 4+ test grades
   - Each student should have 2+ project scores

4. Generate Report:
   Calculate and display:
   - Each student's test average, project average, and final score
   - Letter grade for each student
   - Class average
   - Top 3 performing students
   - Students at risk (final score < 60)

5. Save Results:
   - Save the report to a text file with formatted output

EXPECTED OUTPUT FORMAT:
=== STUDENT GRADE REPORT ===

Student: Alice (ID: S001)
  Test Average: 86.25
  Project Average: 91.50
  Attendance: 90.0%
  Final Score: 87.43
  Letter Grade: B

Class Statistics:
  Class Average: 82.15
  Top 3: Alice (87.43), Bob (85.20), Charlie (83.10)
  At Risk: David (58.50)

Report saved to: student_report.txt

```python
students={}
def add_student(students, student_id, name):
    students[student_id]={
        "name":name,
        "grades":[],
        "attendance":0.0,
        "projects":{}
    }

add_student(students, "S001", "Alice")
students["S001"]["grades"].extend([85, 90, 78])
students["S001"]["attendance"]=0.95
students["S001"]["projects"] = {"proj1": 95, "proj2": 88}

add_student(students, "S002", "Bob")
students["S002"]["grades"].extend([80, 82, 85])      
students["S002"]["attendance"]=0.90
students["S002"]["projects"] = {"proj1": 80, "proj2": 85}

add_student(students, "S003", "Charlie")
students["S003"]["grades"].extend([92, 88, 91])
students["S003"]["attendance"]=0.98
students["S003"]["projects"] = {"proj1": 90, "proj2": 92}

add_student(students, "S004", "David")
students["S004"]["grades"].extend([78, 85, 80])
students["S004"]["attendance"]=0.92
students["S004"]["projects"] = {"proj1": 85, "proj2": 80}

add_student(students, "S005", "Eve")
students["S005"]["grades"].extend([88, 90, 92])
students["S005"]["attendance"]=0.97
students["S005"]["projects"] = {"proj1": 92, "proj2": 95}


def add_grade(students, student_id, grade):
    if student_id in students:
        students[student_id]["grades"].append(grade)
    else:
        print(f"Student ID {student_id} not found.")
        

def calculate_test_average(students, student_id):
    grades=students[student_id]["grades"]
    return sum(grades)/ len(grades) if grades else 0

def calculate_project_average(students, student_id):
    projects = students[student_id]["projects"].values()
    return sum(projects) / len(projects) if projects else 0

def calculate_final_score(students, student_id):
    test_avg = calculate_test_average(students, student_id)
    proj_avg = calculate_project_average(students, student_id)
    attendance = students[student_id]["attendance"] * 100

    return (test_avg * 0.60) + (proj_avg * 0.25) + (attendance * 0.15)

def get_letter_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"
    
import json

def save_to_file(students, filename):
    with open(filename, 'w') as f:
        json.dump(students, f, indent=4)

save_to_file(students, "student.json")

def save_report(report, filename):
    with open(filename, 'w') as f:
        f.write(report)

report = "=== STUDENT GRADE REPORT ===\n\n"

final_scores = []

for sid, data in students.items():
    test_avg = calculate_test_average(students, sid)
    proj_avg = calculate_project_average(students, sid)
    final = calculate_final_score(students, sid)
    grade = get_letter_grade(final)

    final_scores.append((data["name"], final))

    report += f"Student: {data['name']} (ID: {sid})\n"
    report += f"  Test Average: {test_avg:.2f}\n"
    report += f"  Project Average: {proj_avg:.2f}\n"
    report += f"  Attendance: {data['attendance']*100:.1f}%\n"
    report += f"  Final Score: {final:.2f}\n"
    report += f"  Letter Grade: {grade}\n\n"


class_avg = sum(score for _, score in final_scores) / len(final_scores)

top3 = sorted(final_scores, key=lambda x: x[1], reverse=True)[:3]
at_risk = [name for name, score in final_scores if score < 60]

report += "Class Statistics:\n"
report += f"  Class Average: {class_avg:.2f}\n"
report += "  Top 3: " + ", ".join([f"{n} ({s:.2f})" for n, s in top3]) + "\n"

if at_risk:
 report += "  At Risk: " + ", ".join(at_risk) + "\n"

else:
    report += "  At Risk: None\n"

save_report(report, "student_report.txt")
print(report)
```

EXERCISE 2: INVENTORY MANAGEMENT SYSTEM

PROBLEM:
Build an inventory tracking system that monitors stock levels and predicts 
when to reorder products.

REQUIREMENTS:

1. Data Structure:
   Create inventory as a dictionary:
   {
       "PROD001": {
           "name": "Laptop",
           "category": "Electronics",
           "price": 999.99,
           "stock": 50,
           "daily_sales": [5, 3, 7, 4, 6]  # last 5-7 days of sales
       }
   }

2. Functions to Write:
   - add_product(inventory, product_id, name, category, price, stock)
   - record_sale(inventory, product_id, quantity)
     Updates stock and adds to daily_sales list
   - calculate_sales_velocity(inventory, product_id)
     Return average daily sales using the daily_sales list
   - predict_stockout_days(inventory, product_id)
     Calculate days until out of stock: stock / sales_velocity
   - get_low_stock_products(inventory, threshold_days=7)
     Return list of products that will run out within threshold days
   - calculate_category_sales(inventory)
     Return total units sold per category
   - generate_reorder_report(inventory)

3. Create Data:
   - Add at least 8 products across 3 categories (Electronics, Accessories, Office)
   - Each product should have 5-7 days of sales history
   - Vary stock levels (some low, some high)

5. Generate Report:
   Display:
   - Products needing reorder (< 7 days of stock remaining)
   - Sales velocity for each product
   - Total sales by category
   - Products with highest/lowest sales velocity

6. Save Results:
   - Save current inventory state to CSV

EXPECTED OUTPUT FORMAT:
=== INVENTORY REORDER REPORT ===

Products Needing Reorder (< 7 days stock):
  PROD001: Laptop
    Current Stock: 12 units
    Sales Velocity: 4.8 units/day
    Days Until Stockout: 2.5 days
    Recommended Reorder: 50 units

  PROD004: Mouse
    Current Stock: 18 units
    Sales Velocity: 2.9 units/day
    Days Until Stockout: 6.2 days
    Recommended Reorder: 30 units

Category Performance:
  Electronics: 142 units sold
  Accessories: 87 units sold
  Office: 64 units sold

Top Selling Products:
  1. Laptop (velocity: 4.8/day)
  2. USB Cable (velocity: 3.2/day)
  3. Mouse (velocity: 2.9/day)


```python
import csv


# 1. ADD PRODUCT

def add_product(inventory, product_id, name, category, price, stock, daily_sales):
    inventory[product_id] = {
        "name": name,
        "category": category,
        "price": price,
        "stock": stock,
        "daily_sales": daily_sales
    }


# 2. RECORD SALE

def record_sale(inventory, product_id, quantity):
    if product_id in inventory:
        inventory[product_id]["stock"] -= quantity
        inventory[product_id]["daily_sales"].append(quantity)
    else:
        print(f"{product_id} not found!")


# 3. SALES VELOCITY

def calculate_sales_velocity(inventory, product_id):
    sales = inventory[product_id]["daily_sales"]
    return sum(sales) / len(sales) if sales else 0


# 4. STOCKOUT DAYS

def predict_stockout_days(inventory, product_id):
    velocity = calculate_sales_velocity(inventory, product_id)
    stock = inventory[product_id]["stock"]
    return stock / velocity if velocity > 0 else float('inf')


# 5. LOW STOCK PRODUCTS

def get_low_stock_products(inventory, threshold_days=7):
    low_stock = []
    for pid in inventory:
        days = predict_stockout_days(inventory, pid)
        if days < threshold_days:
            low_stock.append(pid)
    return low_stock


# 6. CATEGORY SALES

def calculate_category_sales(inventory):
    category_sales = {}
    for data in inventory.values():
        category = data["category"]
        total_sales = sum(data["daily_sales"])

        category_sales[category] = category_sales.get(category, 0) + total_sales

    return category_sales


# 7. GENERATE REPORT

def generate_reorder_report(inventory):
    report = "=== INVENTORY REORDER REPORT ===\n\n"

    low_stock_products = get_low_stock_products(inventory)

    report += "Products Needing Reorder (< 7 days stock):\n"

    for pid in low_stock_products:
        data = inventory[pid]
        velocity = calculate_sales_velocity(inventory, pid)
        days = predict_stockout_days(inventory, pid)

        reorder_qty = int(velocity * 10)  # simple logic

        report += f"  {pid}: {data['name']}\n"
        report += f"    Current Stock: {data['stock']} units\n"
        report += f"    Sales Velocity: {velocity:.2f} units/day\n"
        report += f"    Days Until Stockout: {days:.2f} days\n"
        report += f"    Recommended Reorder: {reorder_qty} units\n\n"

    # Category performance
    report += "Category Performance:\n"
    category_sales = calculate_category_sales(inventory)
    for cat, total in category_sales.items():
        report += f"  {cat}: {total} units sold\n"

    # Top selling products
    velocities = []
    for pid, data in inventory.items():
        velocity = calculate_sales_velocity(inventory, pid)
        velocities.append((data["name"], velocity))

    velocities.sort(key=lambda x: x[1], reverse=True)

    report += "\nTop Selling Products:\n"
    for i, (name, vel) in enumerate(velocities[:3], start=1):
        report += f"  {i}. {name} (velocity: {vel:.2f}/day)\n"

    return report


# 8. SAVE TO CSV

def save_to_csv(inventory, filename):
    with open(filename, "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["Product ID", "Name", "Category", "Price", "Stock", "Avg Sales"])

        for pid, data in inventory.items():
            avg_sales = sum(data["daily_sales"]) / len(data["daily_sales"])
            writer.writerow([
                pid,
                data["name"],
                data["category"],
                data["price"],
                data["stock"],
                round(avg_sales, 2)
            ])


# 9. CREATE DATA

inventory = {}

add_product(inventory, "PROD001", "Laptop", "Electronics", 999.99, 12, [5,3,7,4,6])
add_product(inventory, "PROD002", "Keyboard", "Accessories", 49.99, 40, [2,3,1,4,2])
add_product(inventory, "PROD003", "Monitor", "Electronics", 199.99, 25, [3,2,4,3,5])
add_product(inventory, "PROD004", "Mouse", "Accessories", 25.99, 18, [2,3,2,4,3])
add_product(inventory, "PROD005", "Printer", "Office", 150.00, 10, [1,2,1,2,1])
add_product(inventory, "PROD006", "USB Cable", "Accessories", 10.00, 60, [4,5,3,4,4])
add_product(inventory, "PROD007", "Notebook", "Office", 5.00, 100, [6,7,5,6,7])
add_product(inventory, "PROD008", "Tablet", "Electronics", 300.00, 20, [2,3,4,2,3])


# 10. RUN REPORT

report = generate_reorder_report(inventory)
print(report)


# 11. SAVE FILE

save_to_csv(inventory, "inventory.csv")
```

EXERCISE 3: WEATHER DATA ANALYSIS SYSTEM

PROBLEM:
Process weather data logs and perform statistical analysis with trend detection.

REQUIREMENTS:

1. Data Structure:
   Use list of tuples for weather data:
   [
       ("2024-01-01", 28.5, 18.2, 65, 0.0),  # (date, high_temp, low_temp, humidity, rainfall)
       ("2024-01-02", 30.1, 19.5, 70, 2.5),
       # ... more days
   ]

2. Functions to Write:
   - generate_weather_data(num_days=30)
     Create synthetic weather data with random values in realistic ranges
   - save_weather_to_csv(weather_data, filename)
   - load_weather_from_csv(filename)
   - calculate_daily_average_temp(day_data)
     Return average of high and low temp for a single day
   - find_extreme_days(weather_data)
     Return tuple of (hottest_day, coldest_day, wettest_day)
   - calculate_weekly_summary(weather_data)
     Group by weeks and return stats for each week
   - detect_temperature_trend(weather_data)
     Use NumPy to determine if temperatures are increasing/decreasing
   - count_rainy_days(weather_data, threshold=1.0)
     Count days with rainfall above threshold

3. NumPy Analysis:
   - Convert weather data to NumPy 2D array
   - Calculate mean, std, min, max for each column (temps, humidity, rainfall)
   - Find correlation between humidity and rainfall
   - Identify anomaly days (temp or rainfall > 2 std from mean)
   - Calculate 7-day moving average for temperatures


EXPECTED OUTPUT FORMAT:
=== WEATHER ANALYSIS REPORT ===

Analysis Period: 2024-01-01 to 2024-02-29 (60 days)

Overall Statistics:
  Average High Temperature: 27.3°C
  Average Low Temperature: 18.5°C
  Average Humidity: 68.2%
  Total Rainfall: 145.8mm

Extreme Days:
  Hottest Day: 2024-02-15 (High: 35.2°C)
  Coldest Day: 2024-01-08 (Low: 12.3°C)
  Wettest Day: 2024-01-22 (Rainfall: 48.5mm)

Temperature Trend: Increasing at 0.12°C per day

Rainy Days: 18 days (30.0% of period)

Weekly Summary:
  Week 1 (Jan 01-07): Avg Temp 24.5°C, Rain 12.5mm
  Week 2 (Jan 08-14): Avg Temp 25.8°C, Rain 8.2mm
  Week 3 (Jan 15-21): Avg Temp 26.3°C, Rain 22.1mm


```python
import random
import csv
from datetime import datetime, timedelta
import numpy as np

# 1. Generate synthetic weather data
def generate_weather_data(num_days=30):
    data = []
    start_date = datetime(2024, 1, 1)

    for i in range(num_days):
        date = start_date + timedelta(days=i)

        high_temp = round(random.uniform(25, 35), 1)
        low_temp = round(random.uniform(15, 25), 1)
        humidity = random.randint(50, 90)
        rainfall = round(random.uniform(0, 20), 1)

        data.append((date.strftime("%Y-%m-%d"), high_temp, low_temp, humidity, rainfall))

    return data


# 2. Save to CSV
def save_weather_to_csv(weather_data, filename):
    with open(filename, "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["date", "high", "low", "humidity", "rainfall"])
        writer.writerows(weather_data)


# 3. Load from CSV
def load_weather_from_csv(filename):
    data = []
    with open(filename, "r") as f:
        reader = csv.reader(f)
        next(reader)  # skip header

        for row in reader:
            date, high, low, hum, rain = row
            data.append((date, float(high), float(low), int(hum), float(rain)))

    return data


# 4. Daily average temperature
def calculate_daily_average_temp(day_data):
    return (day_data[1] + day_data[2]) / 2


# 5. Extreme days
def find_extreme_days(weather_data):
    hottest = max(weather_data, key=lambda x: x[1])
    coldest = min(weather_data, key=lambda x: x[2])
    wettest = max(weather_data, key=lambda x: x[4])

    return hottest, coldest, wettest


# 6. Weekly summary
def calculate_weekly_summary(weather_data):
    weeks = []

    for i in range(0, len(weather_data), 7):
        week = weather_data[i:i+7]

        avg_temp = np.mean([(d[1] + d[2]) / 2 for d in week])
        total_rain = sum(d[4] for d in week)

        weeks.append((week[0][0], week[-1][0], avg_temp, total_rain))

    return weeks


# 7. Temperature trend using NumPy
def detect_temperature_trend(weather_data):
    avg_temps = np.array([(d[1] + d[2]) / 2 for d in weather_data])
    days = np.arange(len(avg_temps))

    slope, _ = np.polyfit(days, avg_temps, 1)

    return slope


# 8. Count rainy days
def count_rainy_days(weather_data, threshold=1.0):
    return sum(1 for d in weather_data if d[4] > threshold)


# 9. NumPy Analysis
def numpy_analysis(weather_data):
    arr = np.array([[d[1], d[2], d[3], d[4]] for d in weather_data])

    stats = {
        "mean": np.mean(arr, axis=0),
        "std": np.std(arr, axis=0),
        "min": np.min(arr, axis=0),
        "max": np.max(arr, axis=0)
    }

    # correlation humidity vs rainfall
    corr = np.corrcoef(arr[:, 2], arr[:, 3])[0, 1]

    # anomalies (>2 std)
    mean = stats["mean"]
    std = stats["std"]

    anomalies = []
    for i, row in enumerate(arr):
        if any(abs(row - mean) > 2 * std):
            anomalies.append(weather_data[i][0])

    # moving average (7-day)
    temps = arr[:, 0]  # using high temp
    moving_avg = np.convolve(temps, np.ones(7)/7, mode='valid')

    return stats, corr, anomalies, moving_avg



# MAIN EXECUTION


weather_data = generate_weather_data(60)

save_weather_to_csv(weather_data, "weather.csv")
loaded_data = load_weather_from_csv("weather.csv")

# Stats
avg_high = np.mean([d[1] for d in loaded_data])
avg_low = np.mean([d[2] for d in loaded_data])
avg_humidity = np.mean([d[3] for d in loaded_data])
total_rain = sum(d[4] for d in loaded_data)

hottest, coldest, wettest = find_extreme_days(loaded_data)

trend = detect_temperature_trend(loaded_data)

rainy_days = count_rainy_days(loaded_data)

weekly = calculate_weekly_summary(loaded_data)

stats, corr, anomalies, moving_avg = numpy_analysis(loaded_data)



# PRINT REPORT


print("=== WEATHER ANALYSIS REPORT ===\n")

print(f"Analysis Period: {loaded_data[0][0]} to {loaded_data[-1][0]} ({len(loaded_data)} days)\n")

print("Overall Statistics:")
print(f"  Average High Temperature: {avg_high:.2f}°C")
print(f"  Average Low Temperature: {avg_low:.2f}°C")
print(f"  Average Humidity: {avg_humidity:.2f}%")
print(f"  Total Rainfall: {total_rain:.2f}mm\n")

print("Extreme Days:")
print(f"  Hottest Day: {hottest[0]} (High: {hottest[1]}°C)")
print(f"  Coldest Day: {coldest[0]} (Low: {coldest[2]}°C)")
print(f"  Wettest Day: {wettest[0]} (Rainfall: {wettest[4]}mm)\n")

print(f"Temperature Trend: {'Increasing' if trend > 0 else 'Decreasing'} at {trend:.2f}°C per day\n")

print(f"Rainy Days: {rainy_days} days ({(rainy_days/len(loaded_data))*100:.1f}%)\n")

print("Weekly Summary:")
for i, w in enumerate(weekly, 1):
    print(f"  Week {i} ({w[0]} to {w[1]}): Avg Temp {w[2]:.2f}°C, Rain {w[3]:.2f}mm")

print("\nAdditional Insights:")
print(f"  Humidity-Rainfall Correlation: {corr:.2f}")
print(f"  Anomaly Days: {anomalies}")
```
